```python
29
import pandas as pd

inventory_df = pd.DataFrame({
    'Product': ['Laptop', 'Mouse', 'Keyboard', 'Monitor'],
    'Quantity': [10, 25, 15, 8]
})

stock = pd.Series([100, 200], index=[1, 3])

inventory_df['Stock_Value'] = stock

print(inventory_df)

inventory_df.drop('Stock_Value', axis=1, inplace=True)

print(inventory_df)
```

```python
30
import pandas as pd

# 1. Read copied table
lab_df = pd.read_clipboard()

# 2. Display DataFrame
print(lab_df)

# 3. Display column names
print(lab_df.columns)

# 4. Retrieve one single-word column
print(lab_df['Name'])

# 5. Retrieve column containing a space
print(lab_df['Total Marks'])

# 6. Create smaller DataFrame
small_df = lab_df[['Name', 'Age', 'Total Marks']]

print(small_df)

# 7. Add a new column
small_df['Status'] = 'Pass'

print(small_df)

# 8. Delete the new column
small_df.drop('Status', axis=1, inplace=True)

print(small_df)
```
